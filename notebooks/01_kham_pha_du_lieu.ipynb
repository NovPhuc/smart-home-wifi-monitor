# Notebook 1: Khám phá dữ liệu RSSI

Mục tiêu: kiểm tra phân phối RSSI, số AP, số phòng, và tính chất missing data.

```python
import os, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 4)
plt.rcParams["axes.grid"] = True

ROOT = os.path.abspath("..")
RAW = os.path.join(ROOT, "data", "raw", "wifi_fingerprints.csv")
```

```python
df = pd.read_csv(RAW)
print("Shape:", df.shape)
df.head()
```

```python
metadata_cols = {"timestamp", "session_id", "device_id", "room", "is_anomaly", "anomaly_type", "confidence"}
ap_cols = [c for c in df.columns if c not in metadata_cols]
print("Số AP:", len(ap_cols))
print("Số phòng:", df["room"].nunique())
print("Số session:", df["session_id"].nunique())
```

```python
rssi = df[ap_cols].replace(-100, np.nan)
missing_pct = rssi.isna().mean().sort_values(ascending=False)
print("Tỷ lệ missing trung bình:", missing_pct.mean())
missing_pct.head(10).plot(kind="bar", title="Tỷ lệ missing theo AP")
plt.ylabel("Missing %")
plt.show()
```

```python
valid = rssi.stack()
print("Min RSSI:", valid.min())
print("Max RSSI:", valid.max())
valid.hist(bins=80, title="Phân phối RSSI (bỏ -100)")
plt.xlabel("RSSI (dBm)")
plt.show()
```
